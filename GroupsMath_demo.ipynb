### 5.1. Isomorphy

Two groups $A$ and $B$ are isomorphic if they have the same structure, even if their elements are different or represented in different ways.

Imagine that you have two groups that look completely different. An **isomorphism** is a way of matching the elements of one group with the elements of the other such that the group operation is preserved.

In other words, if two elements combine in a certain way in the first group, their corresponding elements must combine in exactly the same way in the second group.

So, from the point of view of group theory, **isomorphic groups are essentially the same group written in different ways**. We write:

$$A\cong B$$

For example, the group $C_3$ and $A_3$ are isomorphic.

### 5.3. Generators of a group

Every group has a set of elements that generate every other element of the group via the operation of the group. For example, $C_6$ has the elements $e$, $r$, $r^2$, $r^3$, $r^4$ and $r^5$. You can see that the element $r$ can generate every other element in $C_6$ only operating with itself repeatedly.

$$r=r$$
$$r^2=r*r$$
$$r^3=r*r*r$$
$$r^4=r*r*r*r$$
$$r^5=r*r*r*r*r$$
$$e=r*r*r*r*r*r$$

Then, $r$ is a **generator** of $C_6$. The number of elements that a group needs *at least* to be generated is called its **rank**.

$$\text{rank}(C_6)=1$$

Any cyclic group ($C_n$) can be generated using only one generator. Furthermore, every group that can be generated with only one generator is isomorphic with a cyclic group. Mathematically, is written as:

$$\text{rank}(G)=1 \Longleftrightarrow \exists n:G\cong C_n$$

means that you obtain $r^3$ multiplying $r^2$ by the generator $r$.

## **6. Subgroups**

Can there be a group inside another group? To find this we have to know what's is a group mathematically speaking.

### 6.1. Mathematical definition of a group
So far, you have been computing various things with groups, and basically what you have done is perform operations on their elements. Then, a group is made up of a set (let's call it $M$) and an operation (represented as $*$). Therefore, a group $G$ is:

$$G=(M,*)$$

But, the operation must satisfy some axioms:

- **Closure:** First of all, the elements we use to perform operations are those that belong to the set $M$, so if we want to keep performing operations later on, it is logical to require that the result also be in $M$. This is:

$$a,b\in M\:\Longrightarrow a*b\in M$$

- **Identity:** Secondly, as you already know, every group must have its identity element ($e$). This is:

$$\exists!e\in M:\quad a*e=e*a=a$$

- **Inverses:** Also, as you have already experienced before in the code, every element must have its inverse.

$$\forall a\in M,\quad \exists a^{-1}\in M:\quad a*a^{-1}=e$$

- **Associativity:** And lastly, to avoid any headaches, the operation is required to be associative.

$$\forall a,b,c\in M,\quad (a*b)*c=a*(b*c)$$



For example, we can verify the associativity of three elements of $C_6$ using GroupsMath.

```python
a = Element('r',C6)         # Defining "a" as 'r'
b = Element('r^{3}',C6)     # Defining "b" as 'r^{3}'
c = Element('r^{4}',C6)     # Defining "c" as 'r^{4}'

print( (a*b)*c )            # Showing (a*b)*c
print( a*(b*c) )            # Showing a*(b*c)
```

As you see, the result is the same in both cases, $r^2$.

### 6.2. Defining a subgroup

Since you already know what a group is, a subgroup is nothing more than a group inside another. If $S$ is a subgroup of $G$ (i.e. $S$ is inside $G$ and $S$ is itself a group), we write:

$$S\le G$$

Since the subgroup is inside the parent group, associativity is automatically inherited because the operation works the same way for all elements. This means you don't need to check it again.

In fact, the only thing you really need to check is **closure**, ensuring that when you operate any elements of $S$, the result always remains inside $S$.

For example, $C_6$ has the elements $e$, $r$, $r^2$, $r^3$, $r^4$ and $r^5$. You can make the set $T=\{e,r^2,r^4\}$ and operate every element of $T$ with every other element of $T$:

$$e*e=e     \qquad  e*r^2=r^2       \qquad  e*r^4=r^4$$
$$r^2*e=r^2 \qquad  r^2*r^2=r^4     \qquad  r^2*r^4=e$$
$$r^4*e=r^4 \qquad  r^4*r^2=e       \qquad  r^4*r^4=r^2$$

Every result is also in $T$, so the group $(T,*)$ is a subgroup of $C_6$.

$$(T,*)\le C_6$$

> **Note:** Remember that a group is made up of the set ($T$) and the operation ($*$). $T$ is the set of elements and $(T,*)$ is the subgroup.

Logically, every group is a subgroup of itself, and the identity group ($I$, consisting only of the identity element) is also always a subgroup.

$$G\le G    \qquad  I\le G$$



GroupsMath can show you all the subgroups of every group you want. The subgroups of a group `G` are obtained with `G.subgroups()`. To see the elements of all the subgroups of $C_6$ we need to make a `for` loop in Python. If you don't know how to use `for` loops, read the note below.

```python
for S in C6.subgroups():
    print(S.elements)
```

> **Note:** A `for` loop is just a way to repeat an action for every item in a list. In the code above, `C6.subgroups()` gives us a collection of all the subgroups. The loop takes the first subgroup, calls it `S`, and prints its elements. Then, it automatically moves to the next subgroup and does the same, repeating this process until it finishes the entire list.

You can see that the subgroups of $C_6$ consist of the following elements:

$I$ — `['e']`

$C_2$ — `['e', 'r^{3}']`

$C_3$ — `['e', 'r^{2}', 'r^{4}']`

$C_6$ — `['e', 'r', 'r^{2}', 'r^{3}', 'r^{4}', 'r^{5}']`

The group we found previously, $(T,*)$ with $T=\{e,r^2,r^4\}$, was isomorphic to $C_3$. So $C_3$ is a subgroup of $C_6$.

## **7. Direct product**

### 7.1. Definition of the direct product

We have seen that we can find groups inside other groups, but can we combine groups to form new ones? This is the core idea behind the *direct product* of groups. Imagine you have two different groups, let's call them $G$ and $H$. To create a new group using the direct product, you simply pair their elements up. The elements of this new group will look like $(g, h)$, where $g$ comes from $G$ and $h$ comes from $H$. The resulting group is called the **direct product group**.

$$G \times H$$

To do operations in this new group, you just work component by component. You combine the first elements using the operation of group $G$, and the second elements using the operation of group $H$. It is that simple! The identity element is just the pair of both identities, and the inverse is just the pair of both inverses.

$$(g_1,h_1)*(g_2,h_2) = (g_1*g_2,h_1*h_2)$$

Let's make the Cayley graph of $C_2$, this is:

```python
C2.cayley_graph()
```

Now the direct product of two groups in GroupsMath is made with the operator `*`. For example `C2*C2` means $C_2\times C_2$.

```python
G = C2*C2           # Define the direct product group

G.cayley_graph()    # Show the Cayley graph
```
