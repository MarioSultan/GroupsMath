## 1. Creating groups

In GroupsMath there exist a lot of methods to create groups, the more basic one is using the precooked groups that are avaliable at the library `groupsmath.precoocked`. We'll start with the group $C_6$, it elements have the form:

$$ r^k, \text{with } 0\le k < 6 $$

That is: $r^0$, $r$, $r^2$, $r^3$, $r^4$ and $r^5$, where $r^0$ is commonly known as $e$ (the identity element of $C_6$).

To create this group in GroupsMath simply write `C6`, and thats it. Let's visualize its elements!

```python
print(C6.elements)
```

## 3. Visualizing operations and abelianity

It would be great if there were a way to simultaneously view all possible operations between all the elements of the group. Well, such a thing does exist, and it is known as a "Cayley table."

A Cayley table is a matrix-shaped chart where each square represents the result of operating the element of his row with the element of his column.

GroupsMath can do it, do you want to se the Cayley table of $C_6$? Just write:

```python
C6.cayley_table()
```

### Available Precooked Groups List

| Family | Instances Available | Examples & Description |
| :--- | :--- | :--- |
| **Trivial Group** | `I` | The trivial group containing only the identity element `e`. |
| **Cyclic Groups** | `C1` through `C20` | `C5` represents $C_5$, with element names `['e', 'r', 'r^{2}', 'r^{3}', 'r^{4}']`. |
| **Symmetric Groups** | `S1`, `S2`, `S3`, `S4` | `S3` represents $S_3$ (order 6), `S4` represents $S_4$ (order 24). |
| **Alternating Groups** | `A1`, `A2`, `A3`, `A4` | `A4` represents $A_4$ (order 12) with cycle notation. |
| **Dihedral Groups** | `D1` through `D10` | `D4` represents $D_4$ (order 8) with elements `e`, `r`, `s`, ... |
| **Units Groups Modulo $n$** | `U2` through `U20` | `U8` represents $U(8) = \{1, 3, 5, 7\}$, `U10` represents $U(10)$. |
| **Klein Four-Group** | `V4` | Klein 4-group with names `["e", "a", "b", "ab"]`. |
| **Quaternion Group** | `Q8` | Quaternion group of order 8 with names `["1", "-1", "i", "-i", "j", "-j", "k", "-k"]`. |
| **Dicyclic Groups** | `Dic2`, `Dic3` | Dicyclic groups of order $4n$ ($\text{Dic}_2$ has order 8, $\text{Dic}_3$ order 12). |
