Every group has one —and only one— **identity element**, commonly represented as $e$ (but not always), that satisfies:

$$a*e=a \:\text{ for all elements in the group}$$

That is, multiplying any element with the identity results in the same element.

To find the identity element of a group, for example $C_6$, write:

### 4.3. Inverses

Given an element $a$, you can ask "which element should I operate $a$ with to get the identity element?" This element is called the **inverse element** of $a$ (usually written as $a^{-1}$). In GroupsMath is very easy to compute it. We will take $C_6$ as example.

```python
a = Element('r',C6)     # Create the element

b = a.inverse()         # Define the element "b" as the inverse of "a"

print(b)                # Print "b"
```

The reasoning behind this is that $C_6$ has the relation $r^6=e$.

$$r*r^5=r^6=e\:\Rightarrow\:(r)^{-1}=r^5$$

Furthermore, the inverse of an element can be itself; try $r^{3}$.

### 5.1. Isomorphy

Two groups $A$ and $B$ are isomorphic if they have the same structure, even if their elements are different or represented in different ways.

Imagine that you have two groups that look completely different. An **isomorphism** is a way of matching the elements of one group with the elements of the other such that the group operation is preserved.

In other words, if two elements combine in a certain way in the first group, their corresponding elements must combine in exactly the same way in the second group.

So, from the point of view of group theory, **isomorphic groups are essentially the same group written in different ways**. We write:

$$A\cong B$$

For example, the group $C_3$ and $A_3$ are isomorphic.

### 5.3. Generators of a group

Every group has a set of elements that generate every other element of the group via the operation of the group. For example, $C_6$ has the elements $e$, $r$, $r^2$, $r^3$, $r^4$ and $r^5$. You can see that the element $r$ can generate every other element in $C_6$ only operating with itself repeatedly.

$$r=r$$
$$r^2=r*r$$
$$r^3=r*r*r$$
$$r^4=r*r*r*r$$
$$r^5=r*r*r*r*r$$
$$e=r*r*r*r*r*r$$

Then, $r$ is a **generator** of $C_6$. The number of elements that a group needs *at least* to be generated is called its **rank**.

$$\text{rank}(C_6)=1$$

Any cyclic group ($C_n$) can be generated using only one generator. Furthermore, every group that can be generated with only one generator is isomorphic with a cyclic group. Mathematically, is written as:

$$\text{rank}(G)=1 \Longleftrightarrow \exists n:G\cong C_n$$

### 5.4. Cayley graphs

The Cayley graph of a group are is very effective method to see its structure.

The nodes (dots) represents the elements, and the arrows represents which generator you have to multiply the element by for reaching other element. For example, 

$$r^2 \xrightarrow{r} r^3$$

means that you obtain $r^3$ multiplying $r^2$ by the generator $r$.

In GroupsMath, tha Cayley graphs are constructed writing `.cayley_graph()`. Let's visualize the Cayley graph of $C_6$:

```python
C6.cayley_graph()
```

It's cylic, so it can be generated using onli one geneator (shown in red). Now compare it with the Cayley graph of $S_3$:

```python
S3.cayley_graph()
```

It has the same number of elements as $C_6$, but its internal structure is not the same, therefore:

$$C_6\ncong S_3$$
